```python
import math
import matplotlib.pyplot as plt

points = 100
x_values = [i * (2 * math.pi / points) for i in range(points + 1)]


y_values = [math.sin(x) for x in x_values]


fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x_values, y_values, color="darkviolet", linewidth=2.5, label="y = sin(x)")


ax.axhline(0, color='black', linewidth=1) 
ax.grid(True, linestyle="--", alpha=0.5)


ax.set_title("Trigonometric Sine Wave Graph", fontsize=14)
ax.set_xlabel("Angle (Radians)")
ax.set_ylabel("Amplitude")
ax.legend()

plt.tight_layout()
plt.show()
```

```python
import math
import matplotlib.pyplot as plt


points = 300
time_max = 10.0  
decay_constant = 0.5  
frequency = 4.0  


time_values = [i * (time_max / points) for i in range(points + 1)]
displacement_values = []
envelope_upper = []
envelope_lower = []

for t in time_values:
    
    amplitude = math.exp(-decay_constant * t)
    
    
    y = amplitude * math.sin(2 * math.pi * frequency * t)
    
    displacement_values.append(y)
    envelope_upper.append(amplitude)   
    envelope_lower.append(-amplitude)  

fig, ax = plt.subplots(figsize=(9, 5))


ax.plot(time_values, displacement_values, color="mediumblue", linewidth=2, label="Damped Sine Wave")

#
ax.plot(time_values, envelope_upper, color="crimson", linestyle="--", alpha=0.6, label="Decay Envelope ($e^{-\lambda t}$)")
ax.plot(time_values, envelope_lower, color="crimson", linestyle="--", alpha=0.6)


ax.axhline(0, color="black", linewidth=0.8, alpha=0.7)  # Center line
ax.grid(True, linestyle=":", alpha=0.6)
ax.set_title("Damped Sine Wave (Exponential Decay Over Time)", fontsize=13, pad=12)
ax.set_xlabel("Time ($t$)")
ax.set_ylabel("Displacement ($y$)")
ax.set_xlim(0, time_max)
ax.set_ylim(-1.1, 1.1)
ax.legend(loc="upper right")

plt.tight_layout()
plt.show()
```
