```python
import numpy as np
import random

random_list = [random.randint(1, 100) for _ in range(10)]
print("Original List:", random_list)


even_numbers = [num for num in random_list if num % 2 == 0]
print("Even Numbers:", even_numbers)
```

```python
original_text = "Python is fun to learn"
print("Original String:", original_text)


reversed_string = original_text[::-1]
print("Reversed Characters:", reversed_string)


reversed_words = " ".join(original_text.split()[::-1])
print("Reversed Words Only:", reversed_words)
```
